```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
```

```python
df = pd.read_csv('bank.csv')

# Data preprocessing
df['DATE'] = pd.to_datetime(df['DATE'], format='%d-%b-%y', errors='coerce')
df['VALUE DATE'] = pd.to_datetime(df['VALUE DATE'], format='%d-%b-%y', errors='coerce')

numeric_columns = ['WITHDRAWAL AMT', 'DEPOSIT AMT', 'BALANCE AMT']
# for col in numeric_columns:
    # if isinstance(df[col].iloc[0], str):
    #     df[col] = df[col].str.replace(',', '')
    # df[col] = pd.to_numeric(df[col], errors='coerce')
    # df[col] = df[col].fillna(0)

# Plotting
plt.figure(figsize=(12, 6), facecolor='#F0F0F0')
plt.plot(df['DATE'], df['BALANCE AMT'], color='#8f63ee', label='Balance Amount')
plt.xlabel('Date', color='#555555')
plt.ylabel('Balance Amount', color='#555555')
plt.title('Balance Amount Over Time for Account', color='#333333')
plt.grid(True, linestyle='--', alpha=0.5, color='#ced5ce')

plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
plt.gca().xaxis.set_major_locator(mdates.AutoDateLocator())
plt.gcf().autofmt_xdate()

plt.legend(loc='upper left')
plt.tight_layout()
plt.show()
```
